```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")


file_name = "/content/Chennai_AQIBulletins.csv"
df = pd.read_csv(file_name)


df['date'] = pd.to_datetime(df['date'])

df = df.sort_values('date')

print("Dataset Overview:")
print("-" * 50)
print(df.info())
print("\nFirst 5 rows:")
display(df.head())

fig, axes = plt.subplots(3, 1, figsize=(12, 18))
fig.tight_layout(pad=6.0)


axes[0].plot(df['date'], df['Index Value'], color='tomato', alpha=0.7)
axes[0].set_title('Chennai Air Quality Index (AQI) Trend Over Time', fontsize=14, fontweight='bold')
axes[0].set_xlabel('Date')
axes[0].set_ylabel('AQI (Index Value)')

axes[0].axhline(100, color='green', linestyle='--', label='Safe Limit (100)')
axes[0].legend()


sns.countplot(data=df, x='Air Quality', ax=axes[1], palette='viridis',
              order=df['Air Quality'].value_counts().index)
axes[1].set_title('Number of Days by Air Quality Category', fontsize=14, fontweight='bold')
axes[1].set_xlabel('Air Quality Status')
axes[1].set_ylabel('Total Days')
axes[1].tick_params(axis='x', rotation=45)


sns.countplot(data=df, x='Prominent Pollutant', ax=axes[2], palette='magma',
              order=df['Prominent Pollutant'].value_counts().index)
axes[2].set_title('Frequency of Prominent Pollutants', fontsize=14, fontweight='bold')
axes[2].set_xlabel('Pollutant')
axes[2].set_ylabel('Frequency (Days)')

plt.show()
```
